# Drude Model Exploration

This notebook investigates the wavelength-dependent complex
permittivity of a metal using the Drude model.

```python
import numpy as np
import matplotlib.pyplot as plt

from scipy.constants import c
```

## Wavelength range

We investigate wavelengths from 300 nm to 1500 nm.

```python
wavelength_nm = np.linspace(300, 1500, 1000)

wavelength = wavelength_nm * 1e-9
```

```python
omega = 2 * np.pi * c / wavelength
```

```python
epsilon_inf = 9.5
omega_p = 1.37e16
gamma = 1.05e14
```

```python
from src.drude_model import drude_permittivity

epsilon_m = drude_permittivity(
    omega,
    epsilon_inf,
    omega_p,
    gamma
)
```

```python
epsilon_m[:10]
```

```python
epsilon_real = np.real(epsilon_m)
epsilon_imag = np.imag(epsilon_m)
```

```python
epsilon_real[:5]
```

```python
epsilon_imag[:5]
```

```python
plt.figure(figsize=(8, 5))

plt.plot(wavelength_nm, epsilon_real)

plt.xlabel("Wavelength (nm)")
plt.ylabel("Re(ε)")
plt.title("Real Permittivity from Drude Model")

plt.grid()

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(wavelength_nm, epsilon_imag)

plt.xlabel("Wavelength (nm)")
plt.ylabel("Im(ε)")
plt.title("Imaginary Permittivity from Drude Model")

plt.grid()

plt.show()
```
